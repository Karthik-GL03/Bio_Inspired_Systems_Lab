```python
import random

random.seed(10)

requirements = [20, 30, 25]
available_water = 80

POPULATION_SIZE = 20
GENERATIONS = 100
MUTATION_RATE = 0.1

def create_schedule():
    return [
        random.randint(0, available_water)
        for _ in requirements
    ]

def fitness(schedule):
    total_water = sum(schedule)
    shortage = sum(
        max(0, requirements[i] - schedule[i])
        for i in range(len(requirements))
    )
    excess = max(0, total_water - available_water)

    penalty = 1000 * (shortage + excess)
    return total_water + penalty

def select(population):
    candidates = random.sample(population, 3)
    return min(candidates, key=fitness)

def crossover(p1, p2):
    point = random.randint(1, len(requirements) - 1)
    return p1[:point] + p2[point:]

def mutate(schedule):
    schedule = schedule.copy()
    i = random.randrange(len(schedule))
    schedule[i] = random.randint(0, available_water)
    return schedule

population = [
    create_schedule() for _ in range(POPULATION_SIZE)
]

best_schedule = None

for _ in range(GENERATIONS):
    population.sort(key=fitness)

    if best_schedule is None or fitness(population[0]) < fitness(best_schedule):
        best_schedule = population[0].copy()

    new_population = [population[0].copy()]

    while len(new_population) < POPULATION_SIZE:
        p1 = select(population)
        p2 = select(population)
        child = crossover(p1, p2)

        if random.random() < MUTATION_RATE:
            child = mutate(child)

        new_population.append(child)

    population = new_population

print("Best irrigation schedule:")
for i, water in enumerate(best_schedule):
    print(f"Crop {chr(65+i)}: {water} litres")

print("Total water used:", sum(best_schedule), "litres")
print("Water available:", available_water, "litres")
print("Fitness:", fitness(best_schedule))
```
